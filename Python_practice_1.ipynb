```python
revenue = [50, 60, 70]
expenses = [30, 20, 15]
```

```python
for i in range(len(revenue)):
    profit = revenue[i] - expenses[i]
    margin = profit * 100/ revenue[i]
    print(f"margin Q{i+1}: {margin:.2f}")
```

```python
margins = []

for i in range(len(revenue)):
    profit = revenue[i] - expenses[i]
    margin = profit * 100/ revenue[i]
    margins.append(f"margin Q{i+1}: {margin:.2f}")

margins
```

```python
financials = [
      {"revenue" : 50, "expenses" : 30},
      {"revenue" : 60, "expenses" : 20},
      {"revenue" : 70, "expenses" : 40}
      ]
financials
```

```python
for fi in financials:
    profit = fi["revenue"]- fi["expenses"]
    fi['margin'] = profit * 100 / fi["revenue"]

financials
    
```

```python
finance = {
      "Q1": {"revenue" : 50, "expenses" : 30},
       "Q2":{"revenue" : 60, "expenses" : 20},
      "Q3": {"revenue" : 70, "expenses" : 40}
}
finance
```

```python
for quater, data in finance.items():
    margin = (data["revenue"] - data["expenses"]) / data["revenue"] * 100
    print(f"{quater} : {margin:.2f}%")
```
